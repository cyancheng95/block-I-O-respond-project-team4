```python
import pandas as pd

cols = ["Timestamp", "Hostname", "DiskNumber", "Type", "Offset", "Size", "ResponseTime"]
df = pd.read_csv("CAM-02-SRV-lvm3.csv", names=cols)   # 副檔名是 .gz / .zst / .zip 時 pandas 通常會自動判斷壓縮

print("總筆數:", len(df))
print(df["Type"].value_counts())
print(df["Hostname"].unique(), df["DiskNumber"].unique())

# 時間換算：Windows filetime → 一般時間
df["time"] = pd.to_datetime((df["Timestamp"] - 116444736000000000) // 10, unit="us")
print(df["time"].min(), "~", df["time"].max())

# 回應時間：100ns → 毫秒
df["resp_ms"] = df["ResponseTime"] / 10_000
print(df[["Size", "resp_ms"]].describe())
```
